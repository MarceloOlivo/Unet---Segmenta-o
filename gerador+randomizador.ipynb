```python
import os
import random
from PIL import Image, ImageEnhance


input_folder1 = "C:/Users/User/Desktop/UNET/data/train_images"  
input_folder2 = "C:/Users/User/Desktop/UNET/data/train_masks"   
output_folder1 = "C:/Users/User/Desktop/UNET/data/train_images" 
output_folder2 = "C:/Users/User/Desktop/UNET/data/train_masks"  


os.makedirs(output_folder1, exist_ok=True)
os.makedirs(output_folder2, exist_ok=True)


def get_transformations(i):
    random.seed(i)  

    transformations = {
        "rotation": random.choice([0, 90, 180, 270]),
        "flip": random.random() > 0.5,
        
    }
    return transformations


def apply_transformations(image, transformations):
    image = image.rotate(transformations["rotation"])

    if transformations["flip"]:
        image = image.transpose(Image.FLIP_LEFT_RIGHT)

    
    return image


def generate_images(input_folder1, input_folder2, output_folder1, output_folder2, start=0, end=139, num_variants=40):
    file_index = 0  

    for j in range(start, end + 1):
        input_filename1 = f"slice_000_{j:03d}.png" 
        input_filename2 = f"slice_000_{j:03d}.png"  

        input_filepath1 = os.path.join(input_folder1, input_filename1)
        input_filepath2 = os.path.join(input_folder2, input_filename2)

        if os.path.exists(input_filepath1) and os.path.exists(input_filepath2):
            original_image1 = Image.open(input_filepath1)
            original_image2 = Image.open(input_filepath2)

            for i in range(1, num_variants + 1):
                transformations = get_transformations(i)  
                
                transformed_image1 = apply_transformations(original_image1.copy(), transformations)
                transformed_image2 = apply_transformations(original_image2.copy(), transformations)

                output_filename1 = f"slice_{i:03d}_{file_index:03d}.png"
                output_filename2 = f"slice_{i:03d}_{file_index:03d}.png"

                output_filepath1 = os.path.join(output_folder1, output_filename1)
                output_filepath2 = os.path.join(output_folder2, output_filename2)

                transformed_image1.save(output_filepath1)
                transformed_image2.save(output_filepath2)

            print(f"Imagens geradas para slice_000_{file_index:03d} e mask_000_{file_index:03d}")
            file_index += 1 

        else:
            print(f"Arquivo {input_filename1} ou {input_filename2} não encontrado!")

```

```python

generate_images(input_folder1, input_folder2, output_folder1, output_folder2)
```
