```python
import SimpleITK as sitk
import numpy as np
import os
from PIL import Image
```

```python
nrrd_images = "C:/Users/User/Desktop/Programacao/UNET/nrrd_images"
images_output = "C:/Users/User/Desktop/Programacao/UNET/data/train_images"

nrrd_masks = "C:/Users/User/Desktop/Programacao/UNET/nrrd_masks"
mask_output = "C:/Users/User/Desktop/Programacao/UNET/data/train_masks"
```

```python
nrrd_images_data = [f for f in os.listdir(nrrd_images) if f.endswith('.nrrd')]
nrrd_mask_data = [f for f in os.listdir(nrrd_masks) if f.endswith('.nrrd')]
```

```python

if len(nrrd_images_data) == 0:
    print("Nenhum arquivo NRRD encontrado na pasta.")
else:

    for arquivo in nrrd_images_data:
        caminho_imagens = os.path.join(nrrd_images, arquivo)
        try:              
            image = sitk.ReadImage(caminho_imagens)
            image_array = sitk.GetArrayViewFromImage(image)
            print(f"Arquivo {arquivo} lido com sucesso!")
        except Exception as e:
            print(f"Erro ao carregar a máscara {arquivo}: {e}")

if len(nrrd_mask_data) == 0:
    print("Nenhum arquivo NRRD encontrado na pasta.")
else:

    for arquivo in nrrd_mask_data:
        caminho_mascaras = os.path.join(nrrd_masks, arquivo)
        try:              
            masks = sitk.ReadImage(caminho_mascaras)
            masks_array = sitk.GetArrayViewFromImage(masks)
            print(f"Arquivo {arquivo} lido com sucesso!")
        except Exception as e:
            print(f"Erro ao carregar a máscara {arquivo}: {e}")

```

```python
for i, slice in enumerate(image_array):
    if slice.max() != slice.min():
        slice_imagem_padronizado = ((slice - slice.min()) / (slice.max() - slice.min()) * 255).astype(np.uint8)
    else:
        slice_imagem_padronizado = np.zeros_like(slice, dtype=np.uint8)  
        
    img = Image.fromarray(slice_imagem_padronizado)
    img.save(os.path.join(images_output, f"slice_000_{i:03d}.png"))


for i, slice in enumerate(masks_array):
    
    if slice.max() != slice.min():
        slice_mascaras_padronizado = ((slice - slice.min()) / (slice.max() - slice.min()) * 255).astype(np.uint8)
    else:
        slice_mascaras_padronizado = np.zeros_like(slice, dtype=np.uint8)  
 
    mask = Image.fromarray(slice_mascaras_padronizado)
    mask.save(os.path.join(mask_output, f"slice_000_{i:03d}.png"))

print("Imagens e máscaras processadas e salvas com sucesso!")

```
